## Types of Analysis

#### 1. Temporal Analysis

```python
# Step 1. Month-Wise Analysis

month_counts = df['Month'].value_counts().sort_index()
print(month_counts)

# Visualization

import matplotlib.pyplot as plt

month_counts.plot(kind='bar')
plt.title('Observations per Month')
plt.xlabel('Month')
plt.ylabel('Count')
plt.show()
```

```python
# Step 2. Season-wise Analysis

season_counts = df['Season'].value_counts()
print(season_counts)

# Visualization

season_counts.plot(kind='bar')
plt.title('Observations per Season')
plt.xlabel('Season')
plt.ylabel('Count')
plt.show()
```

```python
# Step 3. Time of Day Analysis
```

```python
# checking nulls in Start_Time

df['Start_Time'].isnull().sum()
```

```python
df['Start_Time'].head(10)
```

```python
# Re-convert properly

df['Start_Time'] = pd.to_datetime(df['Start_Time'], format='%H:%M:%S', errors='coerce')
```

```python
df['Start_Time'] = pd.to_datetime(df['Start_Time'], errors='coerce')
```

```python
df['Start_Time'].head()
```

```python
df['Hour'] = df['Start_Time'].dt.hour
```

```python
df['Hour'].isnull().sum()
```

```python
# Count observations by hour

hour_counts = df['Hour'].value_counts().sort_index()
print(hour_counts)

# Visualization

hour_counts.plot(kind='line', marker='o')
plt.title('Observations by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Count')
plt.show()
```

#### 2. SPATIAL ANALYSIS

```python
# Step 1. Location type abalysis

location_counts = df['Location_Type'].value_counts()
print(location_counts)

# Visualization

location_counts.plot(kind='bar')
plt.title('Observations by Location Type')
plt.xlabel('Location Type')
plt.ylabel('Count')
plt.show()
```

```python
Business Insights :

Grassland dominates bird observations, contributing the majority of sightings

Forest shows significantly lower activity, indicating fewer recorded observations

Grasslands may provide better visibility and favorable habitat conditions for bird detection

Indicates grasslands as key biodiversity hotspots for monitoring and conservation
```

#### 3. Species Analysis

```python
# Step 1. Species Diversity

df['Scientific_Name'].nunique()
```

```python
# Step 2. Top Species Analysis

top_species = df['Common_Name'].value_counts().head(10)
print(top_species)

# Visualization

top_species.plot(kind='barh')
plt.title('Top 10 Most Observed Bird Species')
plt.xlabel('Count')
plt.ylabel('Species')
plt.show()
```

```python
# Step 3. Species vs Location
```

```python
# Create pivot table

species_location = pd.crosstab(df['Common_Name'], df['Location_Type'])
species_location.head()
```

```python
# Get top species

top_species_names = df['Common_Name'].value_counts().head(10).index

filtered_data = species_location.loc[top_species_names]
filtered_data
```

```python
# Visualization

filtered_data.plot(kind='bar', stacked=True)
plt.title('Top Species Distribution by Location Type')
plt.xlabel('Species')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```
